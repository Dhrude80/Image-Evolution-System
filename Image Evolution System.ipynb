```python
from PIL import Image
import numpy as np
import random

img = Image.open("sky ground.jpg")
z = np.array(img).astype(np.int16)

for x in range(100):
    x1 = random.choice(range(z.shape[0]))
    x2 = random.choice(range(x1 , z.shape[0]))
    y1 = random.choice(range(z.shape[1]))
    y2 = random.choice(range(y1 , z.shape[1]))
    
    option = random.choice(["1", "2", "3", "4", "5"])
    if option == "1":
        gray = z[ x1 : x2, y1 : y2, :3].mean(axis = 2)
        z[ x1 : x2, y1 : y2, :3] = gray[:, :, None]
    elif option == "2":
        z[ x1 : x2, y1 : y2, :3] =  255-  z[ x1 : x2, y1 : y2, :3]
    elif option == "3":
        z[ x1 : x2, y1 : y2, :3] = np.clip( z[ x1 : x2, y1 : y2, :3] + 20, 0, 255)
    elif option == "4":
        z[ x1 : x2, y1 : y2, :3] = np.clip( z[ x1 : x2, y1 : y2, :3] - 20, 0, 255)
    elif option == "5":
        chance = random.random()
        if chance <= 0.1:
            z[ x1 : x2, y1 : y2, :3] = np.random.randint( 0, 256, (x2 - x1, y2 - y1, 3))
            
z = np.clip(z, 0, 255)
z = z.astype(np.uint8)
new_img = Image.fromarray(z)
new_img.show()
```
